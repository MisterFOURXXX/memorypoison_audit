```python
# %% [markdown]
# # 01 – Synthetic Data Generation (now using the generator for ablation)

# %%
from memorypoison_audit.data_loader import SyntheticDataGenerator
from memorypoison_audit.core.memory_store import MemoryStore
import json, os

gen = SyntheticDataGenerator(seed=42)
facts = gen.generate_facts(num_facts=500)
store = MemoryStore()
session_id = "synthetic_session"

for fact in facts:
    store.add_fact(session_id, fact)
print(f"Inserted {len(facts)} facts into memory store.")

# %%
# Also generate QA pairs for baseline
qa_pairs = gen.generate_qa_pairs(num_pairs=50)
os.makedirs("experiments/data/sessions", exist_ok=True)
with open("experiments/data/sessions/synthetic_qa.json", "w") as f:
    json.dump(qa_pairs, f)
print("Synthetic QA pairs saved to experiments/data/sessions/synthetic_qa.json")
```
